```python
import os
from pdf2image import convert_from_path
from pdf2image.pdf2image import pdfinfo_from_path

base_dir = '/kaggle/input/datasets/ishahzaibkhan/wasif-ali-wasif-guftugu'
output_base = '/kaggle/working/guftagu_images'

CHUNK_SIZE = 50

def process_book(book_num):
    pdf_path = f'{base_dir}/Guftagu {book_num} By Wasif Ali Wasif.pdf'

    if not os.path.exists(pdf_path):
        print(f'Book {book_num} not found.')
        return

    book_folder = os.path.join(output_base, f'Guftagu_{book_num}')
    os.makedirs(book_folder, exist_ok=True)

    done_pages = set(
        int(f.replace('page_', '').replace('.jpg', ''))
        for f in os.listdir(book_folder)
        if f.startswith('page_') and f.endswith('.jpg')
    )

    total = pdfinfo_from_path(pdf_path)['Pages']

    if len(done_pages) == total:
        print(f'Book {book_num} already complete ({total} pages).')
        return

    print(f'Book {book_num}: {total} pages, {len(done_pages)} done, resuming...')

    for start in range(1, total + 1, CHUNK_SIZE):
        end = min(start + CHUNK_SIZE - 1, total)

        if set(range(start, end + 1)).issubset(done_pages):
            continue

        images = convert_from_path(pdf_path, first_page=start, last_page=end)
        for i, image in enumerate(images):
            page_num = start + i
            if page_num not in done_pages:
                image.save(os.path.join(book_folder, f'page_{page_num}.jpg'), 'JPEG')
            image.close()
        del images

    print(f'Book {book_num} done.')

```

```python
for i in range(1,29):
    process_book(i)
```

```python
import shutil

shutil.make_archive('/kaggle/working/guftagu_images', 'zip', output_base)
print('Done: guftagu_all.zip')
```
